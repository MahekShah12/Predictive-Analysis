```python
import math

# Dataset: [Age, Salary]
X = [
    [22, 25000],
    [25, 30000],
    [28, 35000],
    [35, 60000],
    [40, 70000]
]

# Labels
y = ["No", "No", "Yes", "Yes", "Yes"]

# New person
test = [30, 40000]

# K value
k = 3

# Step 1: Calculate Euclidean distance
distances = []

for i in range(len(X)):
    distance = math.sqrt(
        (X[i][0] - test[0])**2 +
        (X[i][1] - test[1])**2
    )
    distances.append((distance, y[i]))

# Step 2: Sort distances
distances.sort()

# Step 3: Select K nearest neighbors
neighbors = distances[:k]

print("Distances:")
for d in distances:
    print(d)

print("\nNearest Neighbors:")
for n in neighbors:
    print(n)

# Step 4: Majority voting
yes = 0
no = 0

for distance, label in neighbors:
    if label == "Yes":
        yes += 1
    else:
        no += 1

# Step 5: Prediction
if yes > no:
    prediction = "Yes"
else:
    prediction = "No"

print("\nPrediction:", prediction)
```
