```python
import math

# [Age, Experience]
X = [
    [22, 1],
    [25, 2],
    [28, 3],
    [32, 5],
    [35, 7]
]

# Salary
y = [25000, 30000, 35000, 45000, 50000]

# New person
test = [30, 4]

k = 3

# Calculate Euclidean distance
distances = []

for i in range(len(X)):
    d = math.sqrt(
        (X[i][0] - test[0])**2 +
        (X[i][1] - test[1])**2
    )
    distances.append((d, y[i]))

# Sort by distance
distances.sort()

# Select K nearest neighbors
neighbors = distances[:k]

print("Nearest neighbors:")
for n in neighbors:
    print(n)

# Take average of their target values
total = 0

for distance, salary in neighbors:
    total += salary

prediction = total / k

print("Predicted Salary:", prediction)
```
